```python
import pandas as pd
import sqlite3
```

```python
import os
os.getcwd() # get current working directory
```

```python
# Conectamos con la base de datos
connection = sqlite3.connect("revistas.db")

# Obtenemos un cursor que utilizaremos para hacer las queries
crsr = connection.cursor()
```

```python
crsr.execute("PRAGMA foreign_keys = ON;") # Activamos las claves foráneas
```

```python
query = '''
 CREATE TABLE Revistas (
     ISSN varchar(255) PRIMARY KEY,
     Titulo varchar(255),
     Periodicidad int,
     Precio_anual float
 );
 '''
crsr.execute(query)
```

```python
query = '''
 CREATE TABLE Suscriptores (
     Id_Suscriptor integer PRIMARY KEY AUTOINCREMENT,
     DNI varchar(255),
     Nombre varchar(255),
     Id_Direccion varchar(255),
     IBAN varchar(255)
 );
 '''
crsr.execute(query)
```

```python
query = '''
 CREATE TABLE Suscripciones (
     Id_Suscripcion varchar(255) PRIMARY KEY,
     ISSN varchar(255),
     Id_Suscriptor varchar(255),
     Fecha_Inicio text,
     Fecha_Final text,
     Foreign Key (ISSN) REFERENCES Revistas(ISSN),
     Foreign Key (Id_Suscriptor) REFERENCES Suscriptores(Id_Suscriptor)
 );
 '''
crsr.execute(query)
```

```python
query = '''
 CREATE TABLE Direccion_Postal (
     Id_Direccion varchar(255) PRIMARY KEY, 
     Municipio varchar(255),
     Codigo_Postal varchar(255)
 );
 '''
crsr.execute(query)
```
