### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv) and assign it to a variable called chipo.

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(url, sep="\t")
chipo
```

```python
chipo_og = chipo.copy()
```

```python
chipo["item_price"] = chipo["item_price"].str.replace("$", "")
chipo["item_price"] = chipo["item_price"].astype(float)
chipo
```

```python
chipo["precio_unit"] = chipo["item_price"] / chipo["quantity"]
```

```python
print(chipo.sort_values(by="item_price", ascending= False))
```

### Step 3. Name of the max valued product

```python
print(*chipo.loc[chipo["precio_unit"].idxmax(), ["item_name", "precio_unit"]])


prods_caros = chipo[chipo["precio_unit"] == chipo["precio_unit"].max()]

print(*prods_caros["item_name"].unique(), "- Precio:", prods_caros["precio_unit"].max())
```

### Step 4. How many products cost more than $10.00?

```python
prods_diez = chipo[chipo["precio_unit"] > 10]

print(prods_diez["item_name"].count())
```

```python
print(chipo[chipo["precio_unit"] > 10].count())
```

```python
print(chipo[chipo["precio_unit"] > 10].shape[0])
```

### Step 4.1: Y cuántos pedidos se han hecho con un producto de más de 10$? Es lo mismo?

```python
print(prods_diez["order_id"].nunique())
```

### Step 4.2: Y cuántos pedidos se han hecho de más de 10$? Es lo mismo?

```python
total_pedido = chipo.groupby("order_id")["item_price"].sum()

cuenta = total_pedido[total_pedido > 10]
print(len(cuenta))

print()
print(total_pedido.sort_values())
```

### Step 4.3: Y en cuántos pedidos se ha pagado más de 10$ por un mismo producto? Es lo mismo?

```python
mas_de_diez = chipo[chipo["item_price"] > 10]

print(mas_de_diez["order_id"].nunique())
```

### Step 7. What was the quantity of the most expensive item ordered? 2 ways

```python
lo_mas_pagado = chipo[chipo["item_price"] == chipo["item_price"].max()]

print(*lo_mas_pagado["quantity"])
```
