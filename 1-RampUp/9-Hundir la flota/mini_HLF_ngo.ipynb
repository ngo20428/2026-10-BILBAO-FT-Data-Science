## Mini Hundir la Flota
#### Primeros pasos
Vamos a empezar a programar trozos de código que nos servirán para el HLF:

* Crear un Tablero como una lista de listas de 10x10 que contenga caracteres espacio: " "
* Insertar dos barcos sustituyendo en el Tablero los espacios en blanco por el caracter "B":
    * Uno de 4 casillas en horizontal: casillas [1][0], [1][1], [1][2], [1][3]
    * Uno de 3 casillas en vertical:   casillas [3][3], [4][3], [5][3]
* Escribir una celda de código separada de las anteriores para simular que el usuario está efectuando un disparo: 
    *  Solicite al usuario dos coordenadas `i,j` mediante dos inputs
    *  Compruebe si esas coordenadas están ocupada por una casilla barco (que haya una en "B")
        * Si hay una "B":
            * Cambiar el contenido de esa posición por una "x"
            * Imprimir por pantalla el texto "Tocado en posición `i,j`"
        * Si hay un " ":
            * Cambiar el contenido de esa posición por una "o"
            * Imprimir por pantalla el texto "Agua"
    * Imprimir por pantalla el Tablero
* Para comprobar que funciona bien la celda de disparar ejecutarla varias veces metiendo distintas coordenadas

### Bonus
* En la celda de disparar, si el usuario acierta a tocar un barco, vuelve a disparar
* Resuélvelo con funciones:
    * Una para solicitar una coordenada válida a la que llamarás dos veces, una par i y otra par j
    * Una `disparar` que devuelva `True` si el usuario ha acertado o `False` si no.
    


```python
import pprint
tablero = [[" " for i in range(0, 10)] for j in range(0, 10)]

barcos_puestos = [
            # Barco de eslora 3
            [(3, 3), (4, 3), (5, 3)],

            # Barco de eslora 4
            [(1, 0), (1, 1), (1, 2), (1, 3)]
        ]

for posiciones in barcos_puestos:
    for fila, columna in posiciones:
        tablero[fila][columna] = "B"

pprint.pprint(tablero)
```

```python
# DISPARO (confiando que el usuario mete lo que se le pide en el input y no cosas raras)
fila = int(input("Introduce la fila (0-9): "))
columna = int(input("Introduce la columna (0-9): "))


if tablero[fila][columna] == "B":
    tablero[fila][columna] = "X"
    print("Le has dado!")

elif tablero[fila][columna] == " ":
    tablero[fila][columna] = "O"
    print("Agua")

else:
    print("Ya habías disparado ahí")

pprint.pprint(tablero)
```

```python
# DISPARO (teniendo en cuenta si el usuario mete algo raro en el input)
while True:
    fila = input("Introduce la fila (0-9): ")
    columna = input("Introduce la columna (0-9): ")

    if fila.isdigit() and columna.isdigit():
        fila = int(fila)
        columna = int(columna)

        if (0 <= fila < 10 and 0 <= columna < 10):
            break
        else:
            print("Las coordenadas tienen que estar entre 0 y 9, que si no disparas fuera, capitán!")
            continue

    else:
        print("Mete solo números del 0 al 9, porfi :)")

if tablero[fila][columna] == "B":
    tablero[fila][columna] = "X"
    print("Le has dado, capitán!")

elif tablero[fila][columna] == " ":
    tablero[fila][columna] = "O"
    print("Agua :(")

else:
    print("Ya habías disparado ahí.")

pprint.pprint(tablero)
```

```python
def coordenadas():
    while True:
        coordenada = input("Introduce una coordenada entre 0 y 9: ")

        if coordenada.isdigit():
            coordenada = int(coordenada)

            if 0 <= coordenada <= 9:
                return coordenada
            else:
                print("Introduce una coordenada entre 0 y 9, que si no disparas fuera ;)")

        else: 
            print("Coordenada no válida. Prueba otra vez metiendo un número del 0 al 9 :)")

```

```python

import pprint

def disparar(tablero):

    acierto = True
    while acierto == True:

        fila = coordenadas()
        columna = coordenadas()

        if tablero[fila][columna] == "B":
            tablero[fila][columna] = "X"
            acierto = True
            print("Le has dado, capitán!")

        elif tablero[fila][columna] == " ":
            tablero[fila][columna] = "O"
            print("Agua :(")
            acierto = False

        else:
            print("Ya has disparado a esta posición.")
            acierto = False
        
        
        pprint.pprint(tablero)
```
