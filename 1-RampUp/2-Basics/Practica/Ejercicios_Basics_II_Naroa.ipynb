* Comprueba sus tipos

## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

$$radianes = grados*(pi/180)$$

````python
import math
````

````python
grados = float(input("Introduce grados: "))
radianes = grados * (math.pi / 180)
print(f"{grados} grados equivalen a {radianes} radianes.")
````

## Ejercicio 7
Crea un programa que recoja dos inputs. Tiene que comprobar si su suma es igual, superior o inferior a 10, cada condición por separado

````python
cosa_1 = float(input("Introduce un número: "))
cosa_2 = float(input("Introduce otro número: "))
suma = cosa_1 + cosa_2
if suma > 10:
    print("La suma es mayor que 10")
elif suma < 10:
    print("La suma es menor que 10")
else:
    print("La suma es igual a 10")
````

1. True and True and False

4. not False and True and not True

Compruébalo después con python

1. False

## Ejercicio 10
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "Lo primero de todo, cómo están los máquinas"

1. Pasarlo todo a mayúsculas
2. Pasarlo todo a minúsculas
3. Pasar a mayúsculas solo las iniciales de las palabras
4. Crea una lista dividiendolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`
6. Elimina las `a` minúsculas

````python
frase = "Lo primero de todo, cómo están los máquinas"

print(frase.upper())
print(frase.lower())
print(frase.title())
lista = frase.split()
print(lista)
frase_1 = frase.replace(",", ";")
print(frase_1)
print(frase.replace("a", ""))
````
