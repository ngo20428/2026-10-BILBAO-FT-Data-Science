## Ejercicio Extra
Programa una calculadora que

* Pregunte qué operación quieres realizar (multiplicación/división/suma/resta/potencia/salir)
* Pida dos números 
* Realice la operación
* Siga funcionando hasta que el usuario elija salir

#### Bonus:
* Que guarde todas las operaciones realizadas y las imprima por pantalla cuando el usuario elija esa opción
* Que sea robusta frente a entradas erróneas del usuario o frente a operaciones como dividir por 0

```python
import math
import pprint
import time
```

```python
salida = False
historial = []

while salida == False:

    operacion = input("¿Qué operación quieres realizar? Multiplicación, división, suma, resta, potencia, historial o salir.").lower().replace("ó", "o")

    if operacion == "historial":
        if historial:
            pprint.pprint(historial)
        else:
            print("Todavía no ha realizado ninguna operación.")
        continue

    elif operacion != "salir":
        if operacion not in ["multiplicacion", "division", "suma", "resta", "potencia"]:
            print("Operación no válida. Inténtalo de nuevo.")
            continue
        else:
            try:
                num1 = float(input("Introduce el primer número:"))
                num2 = float(input("Introduce el segundo número:"))
            except Exception as e:
                print("Por favor, introducelo en númerico :)")
                continue
    else:
        print("Oki doki. Hasta lueguito.")
        break

    match operacion:
        case "multiplicacion":
            result = num1 * num2
            op = "x"
        case "division":
            if num2 != 0:
                result = num1 / num2
                op = "/"
            else:
                result = "No se puede dividir por 0."
                op = "/"
        case "suma":
            result = num1 + num2
            op = "+"
        case "resta":
            result = num1 - num2
            op = "-"
        case "potencia":
            result = num1 ** num2
            op = "^"
        case "salir":
            salida = True
    operacion_realizada = f'{num1} {op} {num2} = {result}'
    historial.append(operacion_realizada)
    print(operacion_realizada)
```
