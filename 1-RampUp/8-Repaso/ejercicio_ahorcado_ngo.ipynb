## Juego del Ahorcado
El funcionamiento es que vaya mostrando por pantalla ```_ _ _ _ _``` las rayas bajas correspondientes al número de letras que tiene la palabra a adivinar. El usuario deberá insertar una letra y, si esa letra existe, se mostrará o mostrarán las letras acertadas ```S _ S _ _```. 

Si falla, se mostrará el número de errores que lleva y, si lleva 6 errores, entonces se mostrará "Fin del juego".
Encapsúlalo en una función.

````python
def ahorcado():

    while True:
        palabra = input("Introduce la palabra a adivinar: ").upper()

        if not palabra.isalpha():
            print("Introduce solo letras, porfi")
        
        else:
            break

    progreso = ["_"] * len(palabra) # Lo ponemos en lista para luego reemplazar la "_" por las letras adivinadas
    print(" ".join(progreso)) # Lo unimos con espacios para que se imprima bonito :)
    print()

    letras_intentadas = set()
    letras_fallidas = set()
    errores = 0
    errores_max = 6

    while errores < errores_max:
        
        letra = input("Introduce una letra porfi :)").upper()

        # Comprobamos si va tutto bene con la letra
        if len(letra) != 1 or not letra.isalpha():
            print("Por favor, introduce solo una letra :) ")
            continue

        if letra in letras_intentadas:
            print(f"Ya habías intentado la letra '{letra}'. Prueba con otra ;)")
            continue

        # Añadimos la letra al historial de intentos
        letras_intentadas.add(letra)

        # Checkeamos si la letra está en la palabra o no
        if letra in palabra:
            print(f"¡Sí está la {letra}!\n")

            for i, hizkia in enumerate(palabra):
                if hizkia == letra:
                    progreso[i] = letra

        else:
            errores = errores + 1
            letras_fallidas.add(letra)
            print(f'Lo siento, la letra {letra} no está en la palabra :(')

        
        print(" ".join(progreso))
        print()
        
        if letras_fallidas: 
            print("Letras que no están:", letras_fallidas)
        
        print(f'Nº de fallos: {errores} de 6.')
        print()

        if "_" not in progreso:
            print("\nUe!! Has adivinado la palabra! :)")
            break

    if errores == errores_max:
        print(f"Fin del juego. No te quedan más intentos.")
        print(f"La palabra era: {palabra}")
    

````

````python
ahorcado()
````
