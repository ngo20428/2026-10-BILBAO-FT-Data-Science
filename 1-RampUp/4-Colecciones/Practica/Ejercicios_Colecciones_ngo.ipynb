## Ejercicio 1
Dada la siguiente lista:
> ```ejer_1 = [1,2,3,4,5]```

Inviértela para que quede de la siguiente manera
> ```ejer_1 = [5,4,3,2,1]```

````python
ejer_1 = [1,2,3,4,5]

ejer_1.sort(reverse = True)
print(ejer_1)
````

Otra forma

````python
ejer_1 = [1,2,3,4,5]
ejer_1_reversed = ejer_1[::-1]

print(ejer_1_reversed)
````

Otra forma

````python
ejer_1 = [1,2,3,4,5]
ejer_1 = sorted(ejer_1, reverse = True)
print(ejer_1)
````

## Ejercicio 3
Crea una lista nueva con todas las combinaciones de las siguientes dos listas:
> ```ejer_3_1 = ["Hola", "amigo"]```
>
> ```ejer_3_2 = ["Que", "tal"]```

Obten el siguiente output:

```['Hola Que', 'Hola tal', 'amigo Que', 'amigo tal']```

````python
ejer_3_1 = ["Hola", "amigo"]
ejer_3_2 = ["Que", "tal"]
ejer_3_lista = []

for i in ejer_3_1:
    for j in ejer_3_2:
        combi = i + " " + j
        ejer_3_lista.append(combi)
print(ejer_3_lista)
````

## Ejercicio 5
Dada la siguiente lista, elimina todos los valores iguales a 3
> ```ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]```

TIP: No intentes eliminar elementos sobre la lista que estás iterando. Haz una copia con `ejer_5.copy()`.

````python
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]

ejer_5_copy = []

for num in ejer_5:
    if num != 3:
        ejer_5_copy.append(num)
print(ejer_5_copy)
````

````python
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]
ejer_5_copia = ejer_5.copy()
for i in ejer_5:
    if i == 3:
        ejer_5_copia.pop(ejer_5_copia.index(i))

print(ejer_5_copia)
````

## Ejercicio 11
Convierte la siguiente lista en un diccionario
> ```ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]```

TIP: Puedes usar dos variables de control en un bucle `for`. Una por cada valor de las tuplas.

````python
ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]

dicc_11 = {}

for clave, valor in ejer_11:
    if clave not in dicc_11:
        dicc_11[clave] = []
    dicc_11[clave].append(valor)

print(dicc_11)
````
