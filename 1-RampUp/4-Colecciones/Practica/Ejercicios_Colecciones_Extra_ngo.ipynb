* Preguntar qué operación quieres realizar (visualizar todos los libros de la biblioteca/buscar libro/añadir libro/eliminar libro por título)
    * En caso de que pida el usuario elija visualizar se visualizarán todos los libros
    * Buscar libro: búsqueda por título que introduce el usuario y se visualiza todos los campos del libro
    * Añadir libro: se debe solicitar al usuario que introduzca título y autor para el nuevo libro y se añade el libro a la lista de libros
    * Eliminar libro: se solicita al usuario el título del libro y éste debe ser eliminado de la lista de libros
    * Alquilar libro: se solicita al usuario el título del libro y se debe pasar el campo "Alquilado" a True
    * Devolver libro: se solicita al usuario el título del libro y se debe pasar el campo "Alquilado" a False

```python
salida = False
encontrado = False

while salida == False:

    operacion = input("Qué operación desea realizar? Pulse: 1- Visualizar todos los libros de la biblioteca 2- Buscar libro " \
    "3- Añadir libro 4- Eliminar libro 5- Alquilar libro 6- Devolver libro 7- Salir")

    if operacion in ("2", "3", "4", "5", "6"):
        titulo = input("Introduzca el título del libro:")
    elif operacion == "1":
        print("Listado de libros de la biblioteca:")
        print("------------------------------------")
    elif operacion not in ("1", "2", "3", "4", "5", "6", "7"):
        print("Porfi, introduce un número del 1 al 7 :)")
        continue
    else:
        print("Oki doki. Hasta lueguito!")
        salida = True

    match operacion:
        case "1": # Visualizar todos los libros
            for libro in libros:
                print(f"{libro['Titulo']}, {libro['Autor']} - Alquilado: {libro['Alquilado']}")
        
        case "2": # Buscar libro
            for libro in libros:
                if titulo == libro["Titulo"].lower():
                    print(f'Lo tenemos: {libro}')
                    encontrado = True
                    break
            else:
                print(f'Lo siento, no disponemos de {titulo} en la biblioteca.')
        
        case "3": # Añadir libro
            autor = input("Introduzca también el autor del libro:")
            nuevo_libro = {"Titulo": titulo, "Autor": autor, "Alquilado": False}
            libros.append(nuevo_libro)
            print(f"Libro '{titulo}' de {autor} añadido correctamente a la biblioteca!")
    
        case "4": # Eliminar libro
            for libro in libros:
                if titulo == libro["Titulo"].lower():
                    libros.remove(libro)
                    print(f'{libro["Titulo"]} eliminado de la biblioteca :(')
                    encontrado = True
                    break
            else:
                print(f'Lo siento, no es posible eliminar {titulo} dado que no se encuentra en la biblioteca.')
        
        case "5": # Alquilar libro
            for libro in libros:
                if titulo == libro["Titulo"].lower():
                    encontrado = True
                    if libro["Alquilado"] == False:
                        libro["Alquilado"] = True
                        print(f'Disfrute de {libro["Titulo"]}!')
                        break
                    else:
                        print(f'Lo siento, {libro["Titulo"]} ya está alquilado.')
                        break
            else:
                print(f'Lo siento, no disponemos de {titulo} en la biblioteca.')
        
        case "6": # Devolver libro
            for libro in libros:
                if titulo == libro["Titulo"].lower() and libro["Alquilado"] == True:
                    encontrado = True
                    libro["Alquilado"] = False
                    print(f'¡Gracias! Esperamos que haya disfrutado de {libro["Titulo"]}')
                    break
            else:
                print(f'Uy, parece que hay algún error con {titulo}. Por favor, compruebe si ha introducido el título correctamente :)')
        
        case "7": # Salir
            salida = True
```
