## Ejercicio 1
Somos una empresa que tiene varias tiendas de electrodomésticos. Necesitamos un programa para manejar las tiendas, ver las ventas que han tenido, cuántos empleados hay, etc... Para ello vamos a modelizar la tienda en una clase, que tendrá las siguientes características:
* Nombre clase: "Tienda"
* Atributos comunes:
    * Tipo: "Electrodomésticos"
    * Abierta: True
* Atributos propios de cada tienda:
    * Nombre - String
    * Dirección - String
    * Número de empleados - int
    * Ventas últimos 3 meses - Lista de 3 numeros
* Método para calcular las ventas de todos los meses, que devuelva un número con todas las ventas.
* Método que calcula la media de ventas de los últimos meses, por empleado
* Método que devuelve en un string el nombre de la tienda, junto con su dirección.
* Método que obtiene las ventas del último mes.
* Método para dar una proyección de las ventas en caso de haber invertido X dinero en marketing en esa tienda. Siendo X el parámetro de entrada. Si X es menor de 1000, las ventas de los ultimos 3 meses hubiesen sido de (1.2 x ventas), si es mayor o igual de 1000, las ventas hubiesen sido de (1.5 x venta). El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve las ventas con el aumento.

Se pide:
1. Implementar la clase "Tienda"
2. Crear tres tiendas con datos inventados
3. Comprueba en al menos una de ellas, todo lo implementado en la clase tienda (sus atributos, media de ventas, ventas/empleado...)
3. Calcular las ventas del último mes de todas las tiendas. Para ello usa el bucle `for`
4. Imprime por pantalla los nombres de las tiendas cuya dirección lleve el string "Avenida"
5. Seguro que ya has hecho este apartado... Documenta la clase :)

```python
class Tienda:
    '''
    Primero definimos los atributos comunes, que son: tipo y que la tienda está abierta
    '''
    tipo = "Electrodomésticos"
    abierta = True


    def __init__(self, nombre, direccion, nempleados, ventas_3m):
        ''' Establecemos los atributos propios de cada tienda'''
        self.nombre = nombre
        self.direccion = direccion
        self.nempleados = nempleados
        self.ventas_3m = ventas_3m

    
    def ventas_totales(self):
        ''' Definimos las ventas totales, que es la suma de las ventas de los 3 últimos meses.
            Es decir, tenemos que sumar los elementos de la lista.'''
        return sum(self.ventas_3m)

    def med_ventas_empleado(self):
        ''' La media de ventas por empleado: cogemos las ventas totales y las dividimos por empleado '''
        return sum(self.ventas_3m) / self.nempleados

    def datos_tienda(self):
        ''' Datos de la tienda: El nombre y la dirección'''
        return self.nombre + " - " + self.direccion

    def ventas_ult_mes(self):
        ''' Ventas del último mes: es el último elemento de la lista de ventas'''
        return self.ventas_3m[-1] # Seleccionamos el último elemento de la lista de ventas

    def proyeccion_ventas(self, dinero_marketing):
        ''' Proyección de ventas: depende de cuanto se invierte en un nuevo argumento (dinero_marketing). 
            Definimos el if, y reescribimos el valor de la variable ventas (y lo devolvemos).'''
        if dinero_marketing < 1000: # Como hay que reescribir el valor de las ventas de los últimos 3 meses, establecemos un multiplicador
            multiplicador = 1.2
        else:
            multiplicador = 1.5

        self.ventas_3m = [venta * multiplicador for venta in self.ventas_3m] # actualizamos la lista multiplicando cada elemento por el multiplicador
        return self.ventas_3m
```

```python
# IMPRIMIMOS LA DOCUMENTACIÓN
print("DOCUMENTACIÓN:")
print(Tienda.__doc__)
print(Tienda.__init__.__doc__)
print(Tienda.ventas_totales.__doc__)
print(Tienda.med_ventas_empleado.__doc__)
print(Tienda.datos_tienda.__doc__)
print(Tienda.ventas_ult_mes.__doc__)
print(Tienda.proyeccion_ventas.__doc__)
```

```python
# Creación de las tiendas

tienda_1 = Tienda("ABC Elec", "Avenida ABC", 12, [12000, 8000, 10000])
tienda_2 = Tienda("Electros SL", "Calle Qwerty", 3, [7000, 8500, 9750])
tienda_3 = Tienda("Mueblespatucasita", "Qazqaz 987", 25, [18000, 25000, 24500])
```

```python
print("Datos generales:", tienda_1.nombre, "-", tienda_1.tipo, "-", tienda_1.direccion, "- Abierta:", tienda_1.abierta)
print("Ventas últimos 3 meses:", tienda_1.ventas_3m)
print( "Ventas totales: ", tienda_1.ventas_totales())
print("Media ventas por empleado:", tienda_1.med_ventas_empleado())
print(tienda_1.datos_tienda())
print("Ventas último mes: ", tienda_1.ventas_ult_mes())
print("Proyección de ventas: ", tienda_1.proyeccion_ventas(1100))
```

```python
# Ventas del último mes de todas las tiendas
tiendas = [tienda_1, tienda_2, tienda_3]

for tienda in tiendas:
    print(tienda.nombre, ":", tienda.ventas_ult_mes())
```

```python
# Tiendas en Avenidas
for tienda in tiendas:
    if "Avenida" in tienda.direccion:
        print(tienda.nombre, " - ", tienda.direccion)
```

## Ejercicio 2
En este ejercicio vamos a implementar una clase *Perro* en Python. La clase tiene las siguientes características:
* Cosas que sabemos seguro que tiene un perro
    * Tiene 4 patas
    * 2 orejas
    * 2 ojos
    * Una velocidad de 0. Por defecto, el perro está parado
* Cuando se inicialice:
    * El perro será de una determinada raza
    * Por defecto tendrá pelo "Marrón", a no ser que se diga lo contrario.
    * Por defecto no tendrá dueño, a no ser que se diga lo contrario.
    
* Dispondrá también de un método llamado andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro.
* Necesita otro método (parar), donde pondremos la velocidad a 0.
* Otro método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada.


Se pide:
* Implementa la clase *Perro*
* Crea un objeto de tipo *Perro*, sin dueño
* Comprueba que están bien todos sus atributos
* Prueba que ande, y comprueba su velocidad
* Páralo
* Documenta la clase *Perro*

```python
class Perro:
    '''Los atributos seguros de un perro: patas, orejas, ojos, velocidad'''
    patas = 4
    orejas = 2
    ojos = 2
    velocidad = 0

    # Características al crear un perro
    def __init__(self, raza, pelo = "Marrón", dueño = None): # por defecto tiene pelo marrón y no tiene dueño
        ''' Características al crear un perro 
        Por defecto tiene pelo marrón y no tiene dueño'''
        self.raza = raza
        self.pelo = pelo
        self.dueño = dueño

    def andar(self, aumento_velocidad):
        '''Andar: ponemos un argumento que sea el aumento de velocidad.
        Definimos la velocidad como la antigua velocidad + el aumento (que es el argumento de la función)'''
        self.velocidad = self.velocidad + aumento_velocidad # la nueva velocidad es la antigua + el aumento

    def parar(self):
        ''' Con la función parar, establecemos la velocidad a 0.'''
        self.velocidad = 0

    def ladrar(self, entrada):
        ''' Hacemos que imprima el texto que introducimos.'''
        return "GUAU! " + entrada
```

```python
# IMPRIMIMOS LA DOCUMENTACIÓN
print("DOCUMENTACIÓN:")
print(Perro.__doc__)
print(Perro.__init__.__doc__)
print(Perro.andar.__doc__)
print(Perro.parar.__doc__)
print(Perro.ladrar.__doc__)
```

```python
perrito = Perro("labrador")

print("Raza: ", perrito.raza)
print("Pelo: ", perrito.pelo)
print("Dueño: ", perrito.dueño)
print("Patas: ", perrito.patas)
print("Orejas: ", perrito.orejas)
print("Ojos: ", perrito.ojos)
print("Velocidad: ", perrito.velocidad)
```

```python
perrito.andar(10)
print(perrito.velocidad)

perrito.andar(2)
print(perrito.velocidad)

perrito.parar()
print(perrito.velocidad)
```

```python
perrito.ladrar("Hola!")
```
